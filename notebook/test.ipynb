```python
#cell 14

def normalize_text(value):
    if pd.isna(value):
        return ""
    
    text = str(value)
    
    text = text.replace("\xa0", " ")
    
    text = text.strip().lower()
    
    text = re.sub(
        r"\s+",
        " ",
        text
    )
    
    return text
```

```python
#cell 15

def normalize_question(value):
    text = normalize_text(value)
    
    text = text.replace(
        "bawahan saya",
        "saya"
    )
    
    text = text.replace(
        "bawahan",
        "saya"
    )
    
    return text
```

```python
#cell 16

question_comparison = []

for i, (self_col, superior_col) in enumerate(
    zip(self_question_cols, superior_question_cols),
    start=1
):
    
    self_normalized = normalize_question(self_col)
    superior_normalized = normalize_question(superior_col)
    
    question_comparison.append({
        "question_no": i,
        "self_question": self_col,
        "superior_question": superior_col,
        "self_normalized": self_normalized,
        "superior_normalized": superior_normalized,
        "match": self_normalized == superior_normalized
    })

question_comparison_df = pd.DataFrame(
    question_comparison
)

question_comparison_df
```

```python
# ============================================================
# CELL 28 — WIDE TO LONG
# ============================================================

def wide_to_long(
    df,
    question_cols,
    comment_cols,
    rater_type
):

    records = []

    # --------------------------------------------------------
    # Validasi jumlah pertanyaan
    # --------------------------------------------------------

    if len(question_cols) != 50:

        raise ValueError(
            f"{rater_type}: jumlah question = "
            f"{len(question_cols)}, seharusnya 50."
        )

    # --------------------------------------------------------
    # Buat lookup indicator
    # --------------------------------------------------------

    indicator_lookup = (
        indicator_master
        .drop_duplicates("indicator_id")
        .set_index("indicator_id")
        .to_dict("index")
    )

    print(
        f"{rater_type}: "
        f"{len(indicator_lookup)} indicator tersedia."
    )

    # --------------------------------------------------------
    # Validasi semua mapping
    # --------------------------------------------------------

    missing_mapping = []

    for question_no in range(1, 51):

        indicator_id = question_indicator_map.get(
            question_no
        )

        if indicator_id not in indicator_lookup:

            missing_mapping.append({
                "question_no": question_no,
                "indicator_id": indicator_id
            })

    if missing_mapping:

        print("❌ Mapping yang bermasalah:")

        display(
            pd.DataFrame(missing_mapping)
        )

        raise ValueError(
            f"{rater_type}: ada indicator mapping "
            "yang tidak ditemukan."
        )

    # --------------------------------------------------------
    # Loop setiap submission
    # --------------------------------------------------------

    for row_idx, row in df.iterrows():

        # ----------------------------------------------------
        # Tentukan employee & rater
        # ----------------------------------------------------

        if rater_type == "Self":

            employee_name = row.get("Nama")
            rater_name = row.get("Name")

        else:

            employee_name = row.get(
                "Nama bawahan yang dinilai"
            )

            rater_name = row.get("Name")

        # ----------------------------------------------------
        # Loop 50 pertanyaan
        # ----------------------------------------------------

        for question_no, question_col in enumerate(
            question_cols,
            start=1
        ):

            # ------------------------------------------------
            # Ambil jawaban
            # ------------------------------------------------

            answer = row.get(question_col)

            # Skip jawaban kosong
            if pd.isna(answer):
                continue

            if str(answer).strip() == "":
                continue

            # ------------------------------------------------
            # Ambil indicator ID
            # ------------------------------------------------

            indicator_id = question_indicator_map.get(
                question_no
            )

            if indicator_id is None:

                raise ValueError(
                    f"""
Question mapping tidak ditemukan.

Question No : {question_no}
Question    : {question_col}
Rater       : {rater_type}
"""
                )

            # ------------------------------------------------
            # Ambil indicator master
            # ------------------------------------------------

            indicator_info = indicator_lookup.get(
                indicator_id
            )

            if indicator_info is None:

                raise ValueError(
                    f"""
Indicator ID tidak ditemukan.

Indicator ID : {indicator_id}
Question No  : {question_no}
Question     : {question_col}
Rater        : {rater_type}
Source Row   : {row_idx + 2}
"""
                )

            # ------------------------------------------------
            # Comment
            # ------------------------------------------------

            comment = np.nan

            if question_no <= len(comment_cols):

                comment_col = comment_cols[
                    question_no - 1
                ]

                comment = row.get(
                    comment_col
                )

            # ------------------------------------------------
            # Simpan record
            # ------------------------------------------------

            records.append({

                "source_row":
                    row_idx + 2,

                "submission_id":
                    row.get("ID"),

                "submission_start":
                    row.get("Start time"),

                "submission_completion":
                    row.get("Completion time"),

                "last_modified_time":
                    row.get("Last modified time"),

                "rater_type":
                    rater_type,

                "rater_name":
                    rater_name,

                "employee_name":
                    employee_name,

                "employee_email":
                    (
                        row.get("Email")
                        if rater_type == "Self"
                        else np.nan
                    ),

                "direktorat":
                    row.get("Direktorat"),

                "departemen":
                    row.get("Departemen"),

                "unit":
                    row.get("Unit"),

                "question_no":
                    question_no,

                "question_text_raw":
                    question_col,

                "indicator_id":
                    indicator_id,

                "dimension":
                    indicator_info["dimension"],

                "indicator_no":
                    indicator_info["indicator_no"],

                "indicator":
                    indicator_info["indicator"],

                "answer_text":
                    str(answer),

                "comment":
                    comment
            })

    # --------------------------------------------------------
    # Convert ke DataFrame
    # --------------------------------------------------------

    result = pd.DataFrame(records)

    return result
```

```python
#cell 32

def normalize_name(value):
    
    if pd.isna(value):
        return ""
    
    text = str(value)
    
    text = text.lower().strip()
    
    text = re.sub(
        r"[^a-z0-9\s]",
        " ",
        text
    )
    
    text = re.sub(
        r"\s+",
        " ",
        text
    )
    
    return text.strip()
```

```python
canonical_df["employee_name_key"] = (
    canonical_df["employee_name"]
    .apply(normalize_name)
)

canonical_df[
    [
        "rater_type",
        "employee_name",
        "employee_name_key"
    ]
].head(20)
```

```python
#cell 33

canonical_df["submission_start"] = pd.to_datetime(
    canonical_df["submission_start"],
    errors="coerce"
)

canonical_df["submission_completion"] = pd.to_datetime(
    canonical_df["submission_completion"],
    errors="coerce"
)

canonical_df["last_modified_time"] = pd.to_datetime(
    canonical_df["last_modified_time"],
    errors="coerce"
)
```

```python
#cell 34

employee_summary = (
    canonical_df
    .groupby("rater_type")
    .agg(
        unique_employee=(
            "employee_name_key",
            "nunique"
        ),
        total_response=(
            "answer_text",
            "count"
        )
    )
    .reset_index()
)

employee_summary
```

```python
#cell 35

submission_summary = (
    canonical_df
    .groupby(
        [
            "employee_name_key",
            "rater_type",
            "submission_id"
        ],
        dropna=False
    )
    .agg(
        employee_name=(
            "employee_name",
            "first"
        ),
        completion=(
            "submission_completion",
            "max"
        ),
        last_modified=(
            "last_modified_time",
            "max"
        ),
        answer_count=(
            "answer_text",
            "count"
        )
    )
    .reset_index()
)

submission_summary.head()
```

```python
submission_summary["sort_date"] = (
    submission_summary["last_modified"]
    .fillna(
        submission_summary["completion"]
    )
)

submission_summary = submission_summary.sort_values(
    [
        "employee_name_key",
        "rater_type",
        "sort_date"
    ],
    ascending=[
        True,
        True,
        False
    ]
)

submission_summary["submission_rank"] = (
    submission_summary
    .groupby(
        [
            "employee_name_key",
            "rater_type"
        ],
        dropna=False
    )
    .cumcount()
    + 1
)

submission_summary["is_latest_submission"] = (
    submission_summary["submission_rank"] == 1
)

submission_summary.head(20)
```

```python
#cell 37

canonical_df = canonical_df.merge(
    submission_summary[
        [
            "employee_name_key",
            "rater_type",
            "submission_id",
            "submission_rank",
            "is_latest_submission"
        ]
    ],
    on=[
        "employee_name_key",
        "rater_type",
        "submission_id"
    ],
    how="left"
)
```
